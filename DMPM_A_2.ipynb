```python

import numpy as np
import pandas as pd
np.random.seed(42)
n_samples = 1000
student_ids = [f"STU_{1000 + i}" for i in range(n_samples)]
ages = np.random.randint(18, 25, size=n_samples)
gender = np.random.choice(
    ["Male", "Female", "Non-Binary"], size=n_samples, p=[0.48, 0.48, 0.04]
)
study_hours = np.round(
    np.clip(np.random.normal(loc=15, scale=5, size=n_samples), 2, 40), 1
)
attendance = np.round(np.random.beta(a=7, b=2, size=n_samples) * 100, 1)
parental_edu = np.random.choice(
    ["High School", "Bachelor's", "Master's", "Doctorate"],
    size=n_samples,
    p=[0.3, 0.4, 0.2, 0.1],
)
internet_access = np.random.choice(["Yes", "No"], size=n_samples, p=[0.88, 0.12])
base_score = 30 + (study_hours * 1.2) + (attendance * 0.35)
noise = np.random.normal(loc=0, scale=6, size=n_samples)
exam_scores = np.round(np.clip(base_score + noise, 0, 100), 1)
df = pd.DataFrame({
    "Student_ID": student_ids,
    "Age": ages,
    "Gender": gender,
    "Study_Hours_Per_Week": study_hours,
    "Attendance_Percentage": attendance,
    "Parental_Education": parental_edu,
    "Internet_Access": internet_access,
    "Exam_Score": exam_scores,
})
df.to_csv("student_performance_updated_1000.csv", index=False)
print("Dataset created successfully!")
```

```python
import numpy as np
import pandas as pd
df = pd.read_csv("student_performance_updated_1000.csv")
print("--- Data Overview ---")
print(df.info())
print("\n--- Numerical Summary Statistics ---")
print(df.describe())
print("\n--- Average Exam Score by Parental Education ---")
parental_group = (
    df.groupby("Parental_Education")["Exam_Score"]
    .agg([np.mean, np.std, np.min, np.max])
    .round(2))
print(parental_group)
print("\n--- Correlation Matrix ---")
numeric_cols = ["Age", "Study_Hours_Per_Week", "Attendance_Percentage", "Exam_Score"]
print(df[numeric_cols].corr().round(2))
top_performers = df[
    (df["Exam_Score"] >= 85) & (df["Attendance_Percentage"] >= 90)]
print(f"\nNumber of top performers (Score >= 85 & Attendance >= 90%): {len(top_performers)}")
```
