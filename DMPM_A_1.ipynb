```python
def mean_calculation(list1):
  total_sum=0
  count=0
  for n in list1:
    total_sum+=n
    count+=1
  return(total_sum/count)
mean_ans=mean_calculation([1,2,3,4,5])
print(mean_ans)
```

```python
def median_cal(list2):
  list2.sort()
  n = len(list2)
  if n % 2 == 0:
    median = (list2[n // 2 - 1] + list2[n // 2]) / 2
  else:
    median = list2[n // 2]
  return median
median_sol_odd = median_cal([1, 2, 3, 4, 5])
median_sol_even = median_cal([1, 2, 3, 4, 5, 6])
print(f"Median for odd list: {median_sol_odd}")
print(f"Median for even list: {median_sol_even}")
```
