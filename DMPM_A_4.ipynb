```python
#CORRELATION CO-EFFICIENT RELATIONS
import numpy as np
X = [10, 20, 30, 40, 50]
Y = [15, 25, 35, 45, 55]
correlation = np.corrcoef(X, Y)[0, 1]
print("Correlation Coefficient:", correlation)
```

```python
#CO-VARIANCE ANALYSIS
import numpy as np
X = [10, 20, 30, 40, 50]
Y = [15, 25, 35, 45, 55]
covariance_matrix = np.cov(X, Y)
print("Covariance Matrix:")
print(covariance_matrix)
print("Covariance:", covariance_matrix[0, 1])
```

```python
#MIN-MAX NORMALIZATION
import numpy as np
value = np.array([10, 20, 30, 40, 50])
min = np.min(value)
max = np.max(value)
normalized = (value- min) / (max - min)
print("Min-Max:", normalized)
```

```python
#Z-SCORE
import numpy as np
value = np.array([10, 20, 30, 40, 50])
mean = np.mean(value)
std = np.std(value)
z_score = (value - mean) / std
print("Z-Score:", z_score)
print("Mean:",mean)
print("STD:",std)
```

```python
#DECIMAL SCALING
import numpy as np
data = np.array([10, 20, 30, 40, 50])
max_value = np.max(np.abs(data))
j = len(str(int(max_value)))
normalized = data / (10 ** j)
print("Decimal Scaled:", normalized)
```

```python
#CHI-SQUARE TEST
import numpy as np
from scipy.stats import chi2_contingency
data=np.array([[30,10],[10,20]])
chi2,p_value,dof,expected=chi2_contingency(data)
print("Chi-squared Statistic:",chi2)
print("p-value:",p_value)
print("Degrees of Freedom:",dof)
print("\nExpected Frequencies:")
print(expected)
alpha=0.05
if p_value<alpha:
  print("\nReject null hypothesis")
  print("\nThere is a significant association between gender and result.")
else:
  print("\nThere is no significant association between gender and result.")


```
