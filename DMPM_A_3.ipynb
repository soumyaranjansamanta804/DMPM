```python
#LINE PLOT
import matplotlib.pyplot as plt
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 150, 130, 180, 200]
plt.plot(months, sales, marker="o")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()


```

```python
#BAR PLOT
import matplotlib.pyplot as plt
subjects = ["Math", "Science", "English", "Computer"]
marks = [85, 78, 90, 95]
plt.bar(subjects, marks, color="skyblue")
plt.title("Student Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.show()
```

```python
#HISTOGRAM
import matplotlib.pyplot as plt
marks = [45, 50, 55, 60, 60, 65, 70, 70, 75, 80,
         80, 85, 85, 90, 95, 95, 100]
plt.hist(marks, bins=5, color="green", edgecolor="black")
plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()
```

```python
#SCATTER PLOT
import matplotlib.pyplot as plt
hours = [1, 2, 3, 4, 5, 6, 7, 8]
marks = [40, 45, 50, 55, 65, 70, 80, 90]
plt.scatter(hours, marks, color="red")
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```

```python
#PIE CHART
import matplotlib.pyplot as plT
subjects = ["Python", "Java", "C++", "JavaScript"]
students = [40, 25, 20, 15]
plt.pie(
    students,
    labels=subjects,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Programming Language Preference")
plt.show()
```

```python
#BOX PLOT
import seaborn as sns
import matplotlib.pyplot as plt
marks = [45, 50, 55, 60, 60, 65, 70, 70,
         75, 80, 85, 90, 95, 100, 150]
sns.boxplot(x=marks)
plT.title("Marks Distribution")
plt.show()
```
